```python
#Ques2)Create a function to print sum of first N natural numbers
def sum(n):
    total = 0
    for i in range(1,n+1):
        total+=i
    return total

print(sum(10))
```

```python
#Ques3)Function to reverse a number
def reverse_number(num):
    reverse = 0

    while num > 0:
        digit = num % 10
        reverse = reverse * 10 + digit
        num = num // 10
    return reverse

print(reverse_number(23450))
```

```python
#Ques4)Create a function to count digits in number
def count_digits(num):
    count = 0
    while num > 0:
        num = num // 10
        count += 1
    return count

print(count_digits(12345))
```

```python
#Ques5)Create a function to check a palindrome number
def palindrome(num):
    original = num
    reverse = 0
    while num > 0:
        digit = num % 10
        reverse = reverse*10+digit
        num = num // 10
    if original == reverse:
        print("Palindrome")
    else:
        print("Not Palindrome")

palindrome(121)
```

```python
#Ques6)Create a function to generate a fibonaaci series
def fibonacci(n):
    a = 0
    b = 1
    i = 0
    while(i < n):
        print(a)
        c = a + b
        a = b
        b = c
        i += 1

fibonacci(7)
```

```python
#Ques7)CALCULATOR
print("CALCULATOR")
def add(a,b):
  return a+b
def subtract(a,b):
  return a-b
def multiply(a,b):
  return a*b
def divide(a,b):
  return a/b
print("1. Addition")
print("2. Subtraction")
print("3. Multiplication")
print("4. Division")

choice = int(input("Enter your choice"))
num_1=float(input("Enter your first number:"))
num_2=float(input("Enter your second number:"))

if (choice == 1):
  print(add(num_1,num_2))
elif (choice == 2):
  print(subtract(num_1,num_2))
elif (choice == 3):
  print(multiply(num_1,num_2))
elif (choice == 4):
  print(divide(num_1,num_2))
else:
  print("INVALID CHOICE")
```

```python
#Ques10)Handle division by zero using exception handling
try:
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))

    result = num1 / num2

    print("Result =", result)

except ZeroDivisionError:
    print("Cannot divide by zero")

except ValueError:
    print("Please enter valid numbers")
```
